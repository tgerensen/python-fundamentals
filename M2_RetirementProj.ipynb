```python
import logging
logging.basicConfig(level=logging.DEBUG,force=True)

print('Please choose a mode: future value (1) or # of years to a goal (2)') 
try:
    mode = int(input('>'))
except ValueError:
    print('Please enter 1 or 2')
```

This project is essentially a more advanced version of Module 1, which uses functions and serves as a retirement planner. The user will be able to enter an initial amount, rate, number of years, and amount added each year after the initial, and get a future value.

Then, an option will be added to set a future value, and the program will determine the number of years until that value is reached.

```python
def ask_float(prompt):
    while True:
        try:
            return float(input(prompt))
        except ValueError:
            print("Please enter a number")

def ask_int(prompt):
    while True:
        try:
            return int(input(prompt))
        except ValueError:
            print("Please enter an integer")

def future_value(principal, rate, years, annual_deposit):
    balance = principal
    for i in range(0,years):
        if i != 0:
            balance += annual_deposit
        balance += balance*rate
        logging.debug('year %d: balance = %.2f', i+1, balance)
    return round(balance,2)

def num_years(principal, final, rate, annual_deposit):
    balance = principal
    years = 0
    while balance < final:
        if years != 0:
            balance = balance + annual_deposit
        balance = balance + (balance*rate)
        years = years + 1
        if years > 1000:
            logging.error("Goal not reachable with these numbers.")
            return None
    assert balance >= final, "loop ended but goal not met — logic error"
    return years


    
if mode == 1:
    st_amt = ask_float('Starting Amount? ')
    r = ask_float('Interest Rate? (decimal) ')
    n = ask_int('Number of Years? ')
    annual = ask_float('Annual deposit? ')
    final = future_value(st_amt,r,n,annual)
    print('Final amount is: $'+str(final))
elif mode == 2:
    st_amt = ask_float('Starting Amount? ')
    goal = ask_float('Ending Amount? ')
    r = ask_float('Interest Rate? (decimal) ')
    annual = ask_float('Annual deposit? ')
    years_to_goal = num_years(st_amt,goal,r,annual)
    print('It will take '+str(years_to_goal)+' years to reach the goal.')
else:
    print('Please choose one of the two modes.')
```
