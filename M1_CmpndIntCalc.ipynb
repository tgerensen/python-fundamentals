This project is intended to be a compound interest calculator. The user will input a starting amount, annual interest rate, and a number of years. I will add an option for yearly deposits as well.

```python
balance = float(input('Starting Amount? '))
r = float(input('Interest Rate? (decimal) '))
n = int(input('Number of Years? '))
deposits = input('Yearly Deposits? y/n:').lower()
print('Starting balance: $'+str(balance))
```

Now comes the logic of the calculator, which uses a for loop. Using if statements, I will write two for loops, the second allowing for a yearly deposit after the first year.

```python
if deposits == "n":
    for i in range(0,n):
        balance = balance + (balance*r)
        print("Balance at the end of year "+ str(i+1)+': $'+str(balance))
elif deposits == "y":
    for i in range(0,n):
        if i != 0:
            balance = balance + float(input("Year "+str(i+1)+" deposit:"))
        balance = balance + (balance*r)
        print("Balance at the end of year "+ str(i+1)+': $'+str(round(balance,2)))
```
